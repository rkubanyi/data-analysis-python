```python
# Check if a list has the number 5 next t the number 5
# somewhere in the list
# [1,2,3,4,5] -> true

def has_55(data):
    for i in range(len(data)-1):
        if data[i] == 5 and data[i+1] == 5:
            return True
    return False
```

```python
has_55([1,2,3,4,5,5,10])
```

```python
# Default Parameters

def my_country(name = "USA"):
    print(f"I am from {name}")
```

```python
my_country()
```

```python
# Number of Arguments
# If the number of arguments is unknown, add a * before the parameter name

def sum_nums(n1,n2):
    return sum((n1,n2))
```

```python
sum_nums(5,6,7,8)
```

```python
def sum_nums(*n):
    print(f"n = {n}, and the type of n is {type(n)}")
    return sum(n)
```

```python
sum_nums(1,2,3,4)
```

```python
sum_nums(5,6,7)
```

```python
# Keyword Arguments
# You can also send values with the key = value syntax
# The order of the parameters

def my_kids(child1,child2,child3):
    print(f"my first child is '{child1}', my second child is '{child2}', my third child is '{child3}'")
```

```python
my_kids('Ryan', 'Patrick', 'Peter')
```

```python
my_kids(child2 = 'Alice', child1 = 'Jon', child3 = 'Bob')
```

```python
# Arbitrary Keyword Arguments, **kwargs
# This way the function will recieve a dictionay of args
# And can access the items accordingly

def my_name(**name):
    print(f"dictionary name is {name}, and this is the type {type(name)}")
    print(f"First name is {name.get('fname', ' ')} and the last name is {name.get('lname', ' -')}")
```

```python
my_name(fname = 'Rama')
```
