# LAMBDA — anonymous function, one expression only
square = lambda x: x ** 2
add = lambda x, y: x + y
Lambda
A lambda is just a function without a name, written in one line. That's it.
Normal function:
pythondef square(x):
    return x ** 2
Same thing as a lambda:
pythonsquare = lambda x: x ** 2
```

The syntax is:
```
lambda <parameters> : <expression to return>
It always returns whatever the expression evaluates to — no return keyword needed. It can only be one expression, so no multi-line logic, no if/else blocks (though ternary works).
python# Multiple parameters
add = lambda x, y: x + y
add(3, 4)   # 7

# With ternary
classify = lambda x: "positive" if x > 0 else "non-positive"
classify(-3)   # "non-positive"
```

You *could* assign a lambda to a variable like above, but honestly that's not really the point. The point is that lambdas are meant to be written **inline, on the spot, throwaway** — which is exactly why they pair so well with `map`.

## Map
nums = [1, 2, 3, 4]
result = list(map(lambda x: x * 2, nums))
# result = [2, 4, 6, 8]

`map` runs a function against **every item in a list** (or any iterable) and gives you back the results.
```
map(<function>, <iterable>)

# MAP — applies a function to every item in an iterable, returns a map object
nums = [1, 2, 3, 4]
squared = list(map(lambda x: x ** 2, nums))   # [1, 4, 9, 16]

# FILTER — keeps items where function returns True
evens = list(filter(lambda x: x % 2 == 0, nums))  # [2, 4]

# These are equivalent:
squared_comp = [x ** 2 for x in nums]  # list comp version of map above
