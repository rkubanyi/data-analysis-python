```python
# alignment of data
df1 = DataFrame(np.arange(9).reshape(3,3), columns=list('bcd'),
               index = ['Ohio','Texas','Colorado']
               )
df2 = DataFrame(np.arange(12).reshape(4,3), columns=list('bde'),
               index = ['Utah','Ohio','Texas','Oregon']
              )
```

```python
#Create a DataFrame representing stock prices
data = {
    'Date': pd.date_range(start='2026-01-01', end='2026-01-10'),
    'META': np.random.choice(np.arange(100,300),size=10),
    'TSLA': np.random.choice(np.arange(100,300),size=10),
    'GOOGL': np.random.choice(np.arange(100,300),size=10)
}
data
```

```python
df = DataFrame(data, index=data['Date'])
df
```

```python
df = DataFrame(data, index=data['Date'], columns=['META','TSLA','GOOGL'])
df
```

```python
#Find the date with the highest closing price for each stock
df.idxmax()
```

```python
df.idxmin()
```

```python
df_t = df.T
```

```python
# Find the stock with the minimum and maximum values for a specific date: 1/2/26
df_t['2026-01-02'].idxmax()
```

```python
df_t['2026-01-02'].idxmin()
```
