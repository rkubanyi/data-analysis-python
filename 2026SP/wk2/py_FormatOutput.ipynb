### Formatting Outputs with %, .format(), and f-string methods

```python
word = 'Hello\tCIS310!'
```

```python
#Formatting with %
print("The quik brown %s  jumps over the lazy dog" %'cat')
```

```python
print("The quik brown %s  jumps over the lazy dog" %'moose')
```

```python
word1, word2 = 'fox', 'dog'
print("The quik brown %s  jumps over the lazy %s" %(word1, word2))
```

```python
print('My score is %.2f in CIS' %98.88)
```

```python
# using .format() method
print("I'm using the general purpose language {} to perform data analysis".format('Python'))
```

```python
print("I'm using the general purpose language {!r} to perform data analysis".format('Python'))
```

```python
print("The {1} {0} jumps over the {3} {2}".format('fox','brown','dog','lazy'))
```

```python
gpa = 3.998
print("My GPA is: {}".format(gpa))
```

```python
gpa = 3.998
print("My GPA is: {2f}".format(gpa))
```

```python
# using f-strings
n = "Dr. Abdunabi"
t = 'Spring 2026'
print(f'{n} teaches CIS310 in {t}')
```

```python
# using f-strings
n = "Dr. Abdunabi"
t = 'Spring 2026'
print(f'{n} teaches CIS310 in {t!r}')
```

```python
# numbers
radius = 2.535
area = (radius**2) * 1.1415
area
```

```python
# using the input() function
n = input('Enter a number')
print(f'n = {n}, and the type is {type(n)}')
```

```python
radius = float(input('Enter Radius: '))
area = (radius**2) * 1.1415
print(f'the area is {area:2.2f}')
```

```python
# Python Casting
# casting to int
x = int(1)
y = int(2.8)
z = int('3')
w = int(True)
print(f'{x}, {y}, {z}, {w}')
```

```python
# Python Casting
# casting to float
x = float(1)
y = float(2.8)
z = float('3')
w = float(True)
print(f'{x}, {y}, {z}, {w}')
```
