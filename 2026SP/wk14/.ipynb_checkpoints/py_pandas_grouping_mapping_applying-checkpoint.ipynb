```python
# dataset of customer orders
sample_product = ['T-shirt','Blouse','Jacket','Hoodie']
sample_color = ['Red','Orange','Yellow','Green','Blue']
sample_size = ['XS','S','M','L']

data = {
    
    'Order_ID:': np.random.choice(np.arange(500,5001,10),size=60).astype(int),
    'Product': np.random.choice(sample_product,size=60),
    'Color': np.random.choice(sample_color,size=60),
    'Size': np.random.choice(sample_size,size=60),
    'Date': pd.date_range(start='02-20-2026',periods=60)
}

df = DataFrame(data)
df
```

```python
#groupby


cars = ['Toyota','GMC','Ford','Dodge']

df = DataFrame({
    'brand': np.random.choice(cars, size=20),
    'price': np.random.randint(40000,65000,size=20),
    'Sales_Data': pd.date_range(start='04-01-2026',periods=20)
})

df.set_index('Sales_Data', inplace=True)
df
```

```python
df.groupby('brand').count()
df
```

```python
df.groupby('brand').mean()
```

```python
df['price'].groupby(df['brand']).sum()
```

```python
df['cum_sum'] = df.groupby('brand')['price'].cumsum()
df
```

```python
df.groupby(['k1','k2']).mean()
```
