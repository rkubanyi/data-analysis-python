### For Loops

```python
colors = ['red', 'orange', 'yellow', 'green']
```

```python
colors
```

```python
range(0, 5)
```

```python
for i in range(len(colors)):
    print(colors[i])
```

```python
range(5)
```

```python
for i in colors:
    print(i)
```

```python
#Python's built in enumerate function
for i, c in enumerate(colors):
    print(i, c)
```

```python
for i, c in enumerate(colors, start = 1):
    print(i, c)
```

```python
#Multiple lists without a nested loop
colors = ['red', 'orange', 'yellow', 'green', 'purple']
ratio = [0.2,0.3,0.4,0.2,0.5]
for i in range(len(colors)):
    print(f"{int(ratio[i]*100)}% {colors[i]}")
```

```python
for c in colors:
    print(c)
```

```python
#Loop Breaks
for c in colors:
    if c == 'yellow':  
        break
    print(c)
```

```python
#Loop continue
for c in colors:
    if c == 'yellow':  
        continue
    print(c)
```
