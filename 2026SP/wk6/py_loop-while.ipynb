### Python - While Loops

**The General Form of Python While Loops:**
<<code>> while boolean-test:
    statement/s <\code>

```python
lst = list(range(1,9))
lst
```

```python
n = 0
while n < len(lst):
    print(lst[n])
    n = n + 1
```

```python
while lst:
    print(lst.pop())
```

```python
#break keyword
lst = list(range(10,19))

lst
```

```python
n = 0
while n < len(lst):
    if lst[n] == 15:
        break
    print(lst[n])
    n+=1
```

```python
n = 0
while n < len(lst):
    if lst[n] == 15:
        n+=1
        continue
    print(lst[n])
    n+=1
```
