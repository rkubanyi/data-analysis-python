## Notes
- List Comprehensions make it easy to create lists while performing sophisticated filtering, mapping, and conditional logic on their members

- It is simple, powerful tools that you can use in a wide variety of situations

## List Comprehension Format:
<code> new_list = [expression **for member in iterable**]  </code>

## Using Conditional Logic

<code> new_list = [expression **for member in iterable** (if conditional)]  </code>

<code> new_list = [expression (if conditional) **for member in iterable** ] </code>

```python
lst = []

for letter in 'hello':
    print(letter)
    lst.append(letter)
    
lst
```

```python
# [expression for item in iterable]

lst = [letter for letter in 'hello']

lst
```

```python
squares = [n**2 for n in range(1,10)]
squares
```

```python
# Convert Celcius to Farhenheit

c_lst = [4,10,20,25.5,15.8,0,14.9]

f_lst = [((9/5) * t) + 32 for t in c_lst]

f_lst
```

```python
# Price after tax

txns = [1.09,23.56,57.84,4.56,6.78]
tax_rate = .08

final_prices = [round(p*(1+tax_rate),2) for p in txns]
final_prices
```

```python
# filter data using conditional logic
# new_list = [expression for member in iterable (if conditional)]

lst = [n for n in range(1,16) if n % 2 == 0]

lst
```

```python
odds = [n for n in range(1,16) if n % 2 != 0]
odds
```

```python
s = "CIS310 communicate students with analytical thinking and practical skills"

#fid vowels from the sentence

vowels = [l for l in s if l in 'aeiou']
vowels
```

```python
lst = [n for n in range(16) if n % 2 ==0]
lst
```

```python
words = ['Even' if v % 2 == 0 else 'Odd' for v in range(16)]
words
```

```python
# Nested List comprehensions
squares = [num**2 for num in range(1,11)]
squares
```

```python
# Using set and dictionary
t = (n for n in range(11))
t
```

```python
s = {n**2 for n in range(1,11)}
```

```python
s
```

```python
# Dictionaries are similar but require a key

d = {n:n**2 for n in range(5,16)}
d
```

```python
#Use of list in dictionary Comprehension
cities = ['Austin', 'Tacoma', 'Topeka', 'Sacremento', 'Charlotte']

temps = {city:[t for t in range(40,60,5)] for city in cities}
```

```python
temps
```

```python
from random import randint

temps = {city:randint(40,100) for city in cities}
temps
```
