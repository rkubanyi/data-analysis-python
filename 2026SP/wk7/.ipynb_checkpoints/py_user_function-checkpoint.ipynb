```python
#Function with Parameters
def greet (name):
        '''
        This function greets a person's name as the parameter
        '''
        print(f'Hello, {name}, welcome to CIS310')
```

```python
greet('Ryan')
```

```python
greet('Alice')
```

```python
# Using return
def avg(n1,n2,n3):
    return (n1+n2+n3)/3
```

```python
avg(1,2,3)
```

```python
avg(10,5,12)
```

```python
x,y,z = 5,10,8
print(f"The average of values: {x}, {y}, {z} is {round(avg(x,y,z), 2)}")
```

```python
def full_name(fname,lname):
    return lname + ", " + fname
```

```python
print(full_name("Ryan", "Kubanyi"))
```

```python
def min_two_odd(n1,n2):
    if n1 % 2 != 0 and n2 % 2 != 0:
        return min(n1,n2)
    else:
        return max(n1,n2)
```

```python
min_two_odd(93,105)
```

```python
min_two_odd(6,7)
```

```python
def is_prime(num):
    count = 0
    for n in range(2, num):
        if num%n == 0:
            return str(num) + ' is not prime'
    return str(num) + ' is prime'
            
```

```python
is_prime(27)
```

```python
is_prime(7)
```
